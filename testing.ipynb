```python
import pandas as pd
import numpy as np
from Linear_regression import LinearRegression_GradientDescent
from Linear_regression_optimized import LinearRegression_GradientDescent_optimized

data = pd.read_csv("Boston House Price Data.csv")
data
```

```python
X = data.drop(columns=["PRICE"])
y = data["PRICE"]
X
```

```python
import random

X_test = pd.DataFrame(columns=X.columns)
y_test = pd.Series()

x_20 = int(len(X) * 0.2)

while len(X_test) < x_20:
    idx = random.choice(list(X.index))

    X_test = pd.concat([X_test, X.loc[[idx]]])
    y_test = pd.concat([y_test, y.loc[[idx]]])

    X = X.drop(idx)
    y = y.drop(idx)

print(X.shape)
print(X_test.shape)
```

```python
X_val = pd.DataFrame(columns=X.columns)
y_val = pd.Series()
X2 = X.copy()
y2 = y.copy()

x_20 = int(len(X) * 0.2)

while len(X_val) < x_20:
    idx = random.choice(list(X.index))

    X_val = pd.concat([X_val, X.loc[[idx]]])
    y_val = pd.concat([y_val, y.loc[[idx]]])

    X = X.drop(idx)
    y = y.drop(idx)
```

```python
def mse(y, predictii):
    sum = 0
    nr = 0

    for i in y:
        sum += (i - predictii[nr])**2
        nr += 1

    return sum/len(y)
```

```python
def StandardScaler(X, y):

    for x in X.columns:
        mean_x = np.mean(X[x])
        std_x = np.std(X[x])

        X[x] = (X[x]-mean_x)/std_x

    mean_y = np.mean(y)
    std_y = np.std(y)

    y = (y-mean_y)/std_y

    return X, y
```

```python
X_scaled, y_scaled = StandardScaler(X, y)
X_scaled_val, y_scaled_val = StandardScaler(X_val, y_val)
X_scaled_test, y_scaled_test = StandardScaler(X_test, y_test)

X2_scaled, y2_scaled = StandardScaler(X2, y2)
```

```python
model = LinearRegression_GradientDescent_optimized(learning_rate=0.01, early_stopping=True)
model.fit(X_scaled, y_scaled,
           epochs=200)
```

```python
predictii = model.predict(X_scaled_val)
predictii2 = model.predict(X_scaled)

print(mse(y_scaled, predictii2))
print(mse(y_scaled_val, predictii))
```

```python
predictii = model.predict(X_scaled_test)

score = mse(y_scaled_test, predictii)
score
```
